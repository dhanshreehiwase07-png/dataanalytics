```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataframe = pd.read_csv("zomato-1.csv")
print(dataframe)
```

```python
dataframe
```

```python
def handleRate(value):
    if value == 'NEW' or value == '-':
        return np.nan
    else:
        value = str(value).split('/')[0]
        return float(value)

dataframe['rate'] = dataframe['rate'].apply(handleRate)
print(dataframe.head())
```

```python
dataframe.info()
```

```python
dataframe.head()
```

```python
sns.countplot(x=dataframe['listed_in(type)'])
plt.xlabel("Type of restaurant")
plt.title("Count of Restaurants")
plt.show()
```

```python
grouped_data = dataframe.groupby('listed_in(type)')['votes'].sum()
result = pd.DataFrame({'votes': grouped_data})
plt.plot(result, c='green', marker='o')
plt.xlabel("Restaurant Type")
plt.ylabel("Votes")
plt.title("Votes for Different Types of Restaurants")
plt.show()
```

```python
plt.hist(dataframe['rate'], bins=5)
plt.title("Ratings Distribution")
plt.show()
```

```python
couples_data = dataframe['approx_cost(for two people)']
sns.countplot(x=couples_data)
plt.xlabel("Cost for two people")
plt.ylabel("Count")
plt.title("Approximate Cost for Two People")
plt.show()
```

```python
sns.barplot(x='online_order', y='rate', data=dataframe)
plt.title("Online Orders vs Ratings")
plt.show()
```

```python
pivot_table = dataframe.pivot_table(index='listed_in(type)', columns='online_order', aggfunc='size', fill_value=0)

sns.heatmap(pivot_table, annot=True, cmap='YlGnBu', fmt='d')
plt.title("Heatmap")
plt.xlabel("Online Order")
plt.ylabel("Restaurant Type")
plt.show()
```

```python
plt.figure(figsize=(6,6))
sns.countplot(x='listed_in(type)', hue='online_order', data=dataframe)
plt.title("Online and Offline Orders by Restaurant Type")
plt.show()
```
